```python
##### PS_1
# 定义函数Print_values
def Print_values():
    # 输入a,b,c
    a = int(input("Please enter a: "))
    b = int(input("Please enter b: "))
    c = int(input("Please enter c: "))
    #判断a，b大小
    if (a > b):
        # 若a > b，则
        # 判断b，c大小
        if (b > c):
            # 若b > c，则
            print(a,b,c)
        else:
            # 若b < c，则
            # 判断a，c大小
            if (a > c):
                # 若a > c，则
                print(a,c,b)
            else:
                # 若a < c，则
                print(c,a,b)        
    else:
        # 若a < b，则
        # 判断b，c大小
        if (b > c):
            # 若b > c，则
            print("No Answer")
        else:
            # 若b < c，则
            print(c,b,a)
    return a,b,c
# 运行Print_values函数            
Print_values()
```

```python
##### PS_2
### 2.1
# 加载random
import random

# 创建一个值为0的5行10列矩阵M1
M1 = [[0 for j in range(10)] for i in range(5)]

# 创建一个值为0的10行5列矩阵M2
M2 = [[0 for j in range(5)] for i in range(10)]

# 为M1赋值
# 循环5行
for i in range(5):
    # 循环10列
    for j in range(10):
        # M1[i][j]的值为0-50的整数
        M1[i][j] = round((random.random()*100)/2)
print("The matrix M1 is: ",M1,"\n")

# 为M2赋值
# 循环10行
for i in range(10):
    # 循环5列
    for j in range(5):
        # M2[i][j]的值为0-50的整数
        M2[i][j] = round((random.random()*100)/2)
print("The matrix M2 is: ",M2,"\n")
```

```python
### 2.2
# 定义函数Matrix_multip
def Matrix_multip(A,B):
    # 读取矩阵A的行数
    rows = len(A)
    # 读取矩阵B的列数
    cols = len(B[0])
    # 建立结果矩阵
    res = [[0 for j in range(cols)] for i in range(rows)]
    # 计算乘法
    for i in range(rows):
        for j in range(cols):
            for k in range(len(B)):
                res[i][j] = A[i][k] * B[k][j]
    return res

M3 = Matrix_multip(M1, M2)
print("The answer is: ",M3)
```

```python
##### PS_3
# 先定义一个阶乘函数factorial
def factorial(n):
    res = 1
    for i in range(1, n+1):
        res = res * i
    return res

# 定义函数Pascal_triangle
def Pascal_triangle(k):
    # 创建一个k+1列值为0的数组
    result = [0 for i in range(k+1)]
    for i in range(k+1):
        # 使用formula "k choose in i"计算数值
        result[i] = factorial(k)/(factorial(i)*factorial(k-i))
    return result

# 输出第100行的结果
print("The results of the line 100 are: ")
for number in Pascal_triangle(100):
    print(number)

# 输出第200行的结果
print("The results of the line 200 are: ")
for number in Pascal_triangle(200):
    print(number)
```

```python
##### PS_4
# 定义函数Least_moves
def Least_moves(n):
    # 设定初始值为1
    num = 1
    # 初始步骤为0次
    i = 0 
    # 持续循环
    while True:
        # 若初始值等于输入值
        if (num == n):
            # 结束循环
            break
        else:
            # 初始值double后小于等于输入值，则执行double，并步骤+1
            if num * 2 <= n:
                num = num * 2
                i = i + 1
            # 否则执行add，并步骤+1
            else:
                num = num + 1
                i = i + 1
    return i

x = round(random.random()*100)
print()
ans = Least_moves(x)
print("When the random integer x is: ",x,", the smallest number of moves is: ",ans,sep="")
```

```python
##### PS_5
### 5.1
# 定义函数Find_expression
def Find_expression(n):
    # 创建方案sol和方案数num
    sol = ""
    num = 0
    # 执行3的八次方次循环
    for q in range(3):
        for w in range(3):
            for e in range(3):
                for r in range(3):
                    for t in range(3):
                        for y in range(3):
                            for u in range(3):
                                for i in range(3):
                                    # 设置1-9每个数中间的插入的符号
                                    insert = [q,w,e,r,t,y,u,i]
                                    # 最终结果初始化
                                    exp = "1"
                                    # 对每个空缺进行循环
                                    for o in range(8):
                                        # 当insert值为0时，视为+
                                        if insert[o] == 0:
                                            exp = exp + "+" + str(o+2)
                                        # 当insert值为1时，视为-
                                        elif insert[o] == 1:
                                            exp = exp + "-" + str(o+2)
                                        # 当insert值为2时，视为相连接
                                        else:
                                            exp = exp + str(o+2)
                                    # 当结果的算数运算等于输入值时，记录结果和方案数
                                    if eval(exp) == n:
                                        sol = sol + exp + " = " + str(n) + "\n"
                                        num = num + 1 
    return sol, num
                                        
# 输出1-100的方案list
for x in range(1,101):
    Solution,number = Find_expression(x)
    print("For integer ",x,", the solutions are:\n",Solution,sep="")
```
