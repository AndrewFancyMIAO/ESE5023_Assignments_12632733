```python
# import pandas 
import pandas as pd
# import numpy
import numpy as np
# import matplotlib
from matplotlib import pyplot as plt
# make plots appear and be stored within the notebook
#matplotlib inline
```

```python
# AQI of 4 cities on 10/14/2026
city = ['Beijing', 'Shanghai', 'Guangzhou', 'Shenzhen']
aqi  = [153, 47, 60, 74]

# Create a pandas series
my_data = pd.Series(aqi, index=city)

# Show the series using "city" as the index
my_data

# Plot the series as bars
my_data.plot(kind='bar')
plt.show()
```

```python
# Apply operations and functions
# The values make no sense, just for demonstration

np. log(my_data) / my_data**2
```

```python
# Get access the index object
my_data.index

# Get access the values
my_data.values

# Get values using the index via .loc attribute
my_data.loc['Shenzhen']

my_data.loc[['Shenzhen', 'Beijing']]

# Or by row position via .iloc
my_data.iloc[0]

# .iloc is left inclusive but right exclusive
my_data.iloc[0:2]
```

```python
idk = pd.read_csv("ozone_data.csv")
idk

```

```python
idk0 = idk['Temperature'].idxmin()
print(idk['Temperature'].min(),"\n",idk.loc[idk0,['Month','Day']],sep="")
```

```python
idk1 = idk.loc[(idk['Ozone'] > 40)]
idk1
```

```python
idk_index = idk['Ozone'].idxmin()
print('The highest ozone is',idk.loc[idk_index,['Ozone']],'\nThe date is:\n',idk.loc[idk_index,['Month','Day']],sep="")
```
